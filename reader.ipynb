```python
f = open("sample_lanes.csv")
f.readline()
f.readline()
f.readline()
header = f.readline()
print(header)
line = f.readline()

totals = {}
total = 0
rows_read = 0
skipped = []
highest = 0
highest_deaths_lane = 0
worst_lane = ""
worst_date = ""
l1_latest_date = ""
l1_birds = 0
worst_total_lane = ""
worst_total = 0
while line:
    parts = line.strip().split(",")
    if parts[1] =="":
        line = f.readline()
        continue
    try:
        death = int(parts[4])
    except(ValueError, IndexError):
        skipped.append(f"{parts[1]} {parts[3]}")  
        line = f.readline()
        continue
    birds = None
    try:
        birds = int(parts[9])
    except (ValueError, IndexError):
        pass  
    if death > highest:
        highest = death
        worst_lane = parts[3]
        worst_date = parts[1]
    rows_read+=1

    if parts[3] == "L1" and  parts[1]>l1_latest_date:
        l1_birds = birds
        l1_latest_date = parts[1]

    key = parts[3]
    totals[key] = totals.get(key, 0) + death
    total+=death
    line = f.readline()
f.close()
for lane in totals:
    if totals[lane] > worst_total:
        worst_total = totals[lane]
        worst_total_lane = lane
print(f'There are a total {rows_read} rows')
print(f'The date and lane having bad rows were {skipped}')
print(f'{worst_lane} recorded highest deaths of {highest} on date: {worst_date}')
print(f'There are {l1_birds} on latest date: {l1_latest_date}')
print(f'Deaths in each lanes {totals}')
print(f'The most deaths in a single lane is {worst_total} from lane {worst_total_lane}')
```
